## Reproducible measurements

These cells run the experiments reported in the post. The synthetic datasets have fixed seeds and known targets. The result files preserve individual replications, settings and timings.

```python
"""Post 109 simulations and an explicitly assumed-return GARCH allocation."""
import json
from pathlib import Path
import time

import numpy as np
import pandas as pd

ROOT = WORK
OUT = ROOT / 'scripts/kelly_results'
OUT.mkdir(parents=True, exist_ok=True)
t0 = time.perf_counter()
rng = np.random.default_rng(109)
p,b=0.55,1.0
binary_kelly=p-(1-p)/b
fractions=np.linspace(0,0.5,301)
growth=p*np.log1p(b*fractions)+(1-p)*np.log1p(-fractions)
mu,sigma=0.10,0.20
f_true=mu/sigma**2
years=10
# Exact terminal distribution for constant-fraction rebalancing in a GBM.
estimated_mu=mu+sigma/np.sqrt(5)*rng.normal(size=10000)
estimated_kelly=np.clip(estimated_mu/sigma**2,0,10)
future_z=rng.normal(size=10000)
estimation={}
for scale in [0.25,0.5,1,2]:
    f=scale*estimated_kelly
    log_terminal=(f*mu-0.5*f*f*sigma*sigma)*years+f*sigma*np.sqrt(years)*future_z
    estimation[str(scale)]={'median_terminal':float(np.exp(np.median(log_terminal))),
                            'mean_log_terminal':float(log_terminal.mean())}
days=2500
dt=1/250
increments=mu*dt+sigma*np.sqrt(dt)*rng.normal(size=(10000,days)).astype(np.float32)
fans={}
drawdowns={}
for scale in [0.25,0.5,1,2]:
    f=scale*f_true
    logwealth=np.cumsum(f*increments-0.5*f*f*sigma*sigma*dt,axis=1)
    running=np.maximum.accumulate(np.maximum(logwealth,0),axis=1)
    dd=1-np.exp(np.min(logwealth-running,axis=1))
    drawdowns[str(scale)]=dd.tolist()
    fans[str(scale)]=np.exp(np.quantile(logwealth[:,::25],[0.1,0.5,0.9],axis=0)).tolist()
prices=pd.read_csv(ROOT/'scripts/garch_results/sp500.csv',parse_dates=['date']).set_index('date')['close']
returns=prices.pct_change().shift(-1)
forecasts=pd.read_csv(ROOT/'scripts/garch_results/forecasts.csv',parse_dates=['origin'])
forecasts=forecasts[forecasts.horizon.eq(1)] if 'horizon' in forecasts else forecasts[forecasts.iloc[:,1].eq(1)]
dates=pd.DatetimeIndex(forecasts.origin)
r=returns.reindex(dates).to_numpy()
variance=forecasts.garch.to_numpy()/10000
assumed_premium=0.05
cap=2.0
base_var=np.var(np.diff(np.log(prices.loc[:dates[0]].to_numpy()))[-1260:])
constant=np.repeat(min(cap,assumed_premium/(252*base_var)),len(r))
dynamic=np.clip(assumed_premium/(252*variance),0,cap)
backtest={}
for name,f in [('constant',constant),('garch',dynamic)]:
    turnover=np.abs(np.diff(np.r_[0,f]))
    net=f*r-0.0005*turnover-0.02*np.maximum(f-1,0)/252
    assert np.all(net>-1)
    logwealth=np.cumsum(np.log1p(net))
    drawdown=1-np.exp(logwealth-np.maximum.accumulate(np.maximum(logwealth,0)))
    backtest[name]={'wealth':np.exp(logwealth).tolist(),'exposure':f.tolist(),
                    'cagr':float(np.expm1(logwealth[-1]*252/len(r))),
                    'max_drawdown':float(drawdown.max()),'mean_exposure':float(f.mean()),
                    'turnover':float(turnover.sum())}
result={'p':p,'b':b,'binary_kelly':binary_kelly,'binary_double_growth':float(p*np.log1p(2*binary_kelly)+(1-p)*np.log1p(-2*binary_kelly)),
        'mu':mu,'sigma':sigma,'true_kelly':f_true,'estimation':estimation,
        'estimated_kelly':estimated_kelly.tolist(),'fans':fans,'drawdowns':drawdowns,
        'backtest':backtest,'dates':dates.strftime('%Y-%m-%d').tolist(),
        'assumed_premium':assumed_premium,'leverage_cap':cap,'transaction_cost_bps':5,
        'annual_leverage_financing':0.02,'seconds':time.perf_counter()-t0}
(OUT/'results.json').write_text(json.dumps(result))
print({k:v for k,v in result.items() if k in ['estimation','seconds','binary_double_growth']})
print({name:{k:v for k,v in d.items() if k not in ['wealth','exposure']} for name,d in backtest.items()})
```
